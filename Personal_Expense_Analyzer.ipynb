```python
# 💰 Personal Expense Analyzer

A beginner-friendly Python project to analyze personal expenses.

## Features
- Calculate total expenses
- Calculate category-wise expenses
- Find highest expense
- Find lowest expense
- Calculate average expense
- Find highest spending category
- Generate a final expense report

### Technologies
- Python
- Jupyter Notebook

### Concepts Used
- Variables
- Dictionaries
- Lists
- Loops
- Functions
- `sum()`
- `len()`
- `max()`
- `min()`
```

```python
# Personal Expense Data

expenses = {
    "Food": [500, 250, 300],
    "Transport": [300, 200, 150],
    "Education": [1000, 500],
    "Shopping": [800, 450],
    "Entertainment": [400, 250]
}

print("Expense data created successfully!")
```

```python
# Display all expenses

print("Personal Expenses")
print("-" * 30)

for category, amounts in expenses.items():
    print(f"{category}: {amounts}")
```

```python
# Calculate total expenses

total_expenses = 0

for category, amounts in expenses.items():
    total_expenses += sum(amounts)

print(f"Total Expenses: PKR {total_expenses}")
```

```python
# Calculate total expense for each category

category_totals = {}

for category, amounts in expenses.items():
    category_totals[category] = sum(amounts)

print("Category-wise Expenses")
print("-" * 30)

for category, total in category_totals.items():
    print(f"{category}: PKR {total}")
```

```python
# Find the highest individual expense

highest_expense = 0
highest_category = ""

for category, amounts in expenses.items():
    for amount in amounts:
        if amount > highest_expense:
            highest_expense = amount
            highest_category = category

print(f"Highest Expense: PKR {highest_expense}")
print(f"Category: {highest_category}")
```

```python
# Find the lowest individual expense

lowest_expense = None
lowest_category = ""

for category, amounts in expenses.items():
    for amount in amounts:
        if lowest_expense is None or amount < lowest_expense:
            lowest_expense = amount
            lowest_category = category
print(f"Lowest Expense: PKR {lowest_expense}")
print(f"Category: {lowest_category}")
```

```python
# Calculate the average individual expense

number_of_expenses = 0

for amounts in expenses.values():
    number_of_expenses += len(amounts)

average_expense = total_expenses / number_of_expenses
print(f"Number of Expenses: {number_of_expenses}")
print(f"Average Expense: PKR {average_expense:.2f}")
```

```python
# Find the category where the most money was spent

highest_category = max(category_totals, key=category_totals.get)
highest_category_amount = category_totals[highest_category]

print(f"Highest Spending Category: {highest_category}")
print(f"Amount Spent: PKR {highest_category_amount}")
```

```python
# Find the category where the least money was spent

lowest_category = min(category_totals, key=category_totals.get)
lowest_category_amount = category_totals[lowest_category]


print(f"Lowest Spending Category: {lowest_category}")
print(f"Amount Spent: PKR {lowest_category_amount}")

```
