# 💰 Personal Expense Analyzer

A beginner-friendly Python project to analyze personal expenses.

## Features
- Calculate total expenses
- Calculate category-wise expenses
- Find highest expense
- Find lowest expense
- Calculate average expense
- Find highest spending category
- Generate a final expense report

### Technologies
- Python
- Jupyter Notebook

### Concepts Used
- Variables
- Dictionaries
- Lists
- Loops
- Functions
- `sum()`
- `len()`
- `max()`
- `min()`

```python
# Personal Expense Data

expenses = {
    "Food": [500, 250, 300],
    "Transport": [300, 200, 150],
    "Education": [1000, 500],
    "Shopping": [800, 450],
    "Entertainment": [400, 250]
}

print("Expense data created successfully!")
```

```python
# Display all expenses

print("Personal Expenses")
print("-" * 30)

for category, amounts in expenses.items():
    print(f"{category}: {amounts}")
```

```python
# Calculate total expenses

total_expenses = 0

for category, amounts in expenses.items():
    total_expenses += sum(amounts)

print(f"Total Expenses: PKR {total_expenses}")
```

```python
# Calculate total expense for each category

category_totals = {}

for category, amounts in expenses.items():
    category_totals[category] = sum(amounts)

print("Category-wise Expenses")
print("-" * 30)

for category, total in category_totals.items():
    print(f"{category}: PKR {total}")
```

```python
# Find the highest individual expense

highest_expense = 0
highest_expense_category = ""

for category, amounts in expenses.items():
    for amount in amounts:
        if amount > highest_expense:
            highest_expense = amount
            highest_expense_category = category

print(f"Highest Expense: PKR {highest_expense}")
print(f"Category: {highest_expense_category}")
```

```python
# Find the lowest individual expense

lowest_expense = None
lowest_expense_category = ""

for category, amounts in expenses.items():
    for amount in amounts:
        if lowest_expense is None or amount < lowest_expense:
            lowest_expense = amount
            lowest_expense_category = category
print(f"Lowest Expense: PKR {lowest_expense}")
print(f"Category: {lowest_expense_category}")
```

```python
# Calculate the average individual expense

number_of_expenses = 0

for amounts in expenses.values():
    number_of_expenses += len(amounts)

average_expense = total_expenses / number_of_expenses
print(f"Number of Expenses: {number_of_expenses}")
print(f"Average Expense: PKR {average_expense:.2f}")
```

```python
# Find the category where the most money was spent

highest_category = max(category_totals, key=category_totals.get)
highest_category_amount = category_totals[highest_category]

print(f"Highest Spending Category: {highest_category}")
print(f"Amount Spent: PKR {highest_category_amount}")
```

```python
# Find the category where the least money was spent

lowest_category = min(category_totals, key=category_totals.get)
lowest_category_amount = category_totals[lowest_category]


print(f"Lowest Spending Category: {lowest_category}")
print(f"Amount Spent: PKR {lowest_category_amount}")
```

```python
# Generate the final expense report

print("=" * 45)
print("        PERSONAL EXPENSE REPORT")
print("=" * 45)

print(f"Total Expenses       : PKR {total_expenses}")
print(f"Number of Expenses   : {number_of_expenses}")
print(f"Average Expense      : PKR {average_expense:.2f}")

print("-" * 45)

print(f"Highest Expense      : PKR {highest_expense}")
print(f"Highest Expense Type : {highest_expense_category}")

print(f"Lowest Expense       : PKR {lowest_expense}")
print(f"Lowest Expense Type  : {lowest_expense_category}")

print("-" * 45)

print(f"Highest Category     : {highest_category}")
print(f"Category Total       : PKR {highest_category_amount}")

print(f"Lowest Category      : {lowest_category}")
print(f"Category Total       : PKR {lowest_category_amount}")

print("=" * 45)

print("\nCategory Breakdown")
print("-" * 45)

for category, total in category_totals.items():
    print(f"{category:<20} PKR {total}")

print("=" * 45)
```

```python
# Basic analysis

print("📊 Expense Analysis")
print("-" * 30)

if highest_category_amount > total_expenses * 0.30:
    print(
f"You are spending a large portion of your money "
        f"on {highest_category}."
    )
else:
    print("Your expenses are reasonably distributed across categories.")
if average_expense > 700:
    print("Your average expense is relatively high.")
else:
    print("Your average expense is relatively low.")
```

```python
# Add a new expense manually

category = input("Enter expense category: ")
amount = float(input("Enter expense amount: PKR "))

if category in expenses:
    expenses[category].append(amount)
else:
    expenses[category] = [amount]

print("\nExpense added successfully!")

print(f"{category}: {expenses[category]}")
```

```python
# Recalculate category totals

category_totals = {}

for category, amounts in expenses.items():
    category_totals[category] = sum(amounts)

# Recalculate total expenses

total_expenses = sum(category_totals.values())

# Recalculate number of expenses

number_of_expenses = sum(len(amounts) for amounts in expenses.values())

# Recalculate average

average_expense = total_expenses / number_of_expenses

# Recalculate highest and lowest individual expenses

highest_expense = 0
highest_expense_category = ""

lowest_expense = None
lowest_expense_category = ""

for category, amounts in expenses.items():
    for amount in amounts:
        if amount > highest_expense:
            highest_expense = amount
            highest_expense_category = category

        if lowest_expense is None or amount < lowest_expense:
            lowest_expense = amount
            lowest_expense_category = category
# Recalculate highest and lowest categories

highest_category = max(category_totals, key=category_totals.get)
highest_category_amount = category_totals[highest_category]

lowest_category = min(category_totals, key=category_totals.get)
lowest_category_amount = category_totals[lowest_category]

print("All calculations updated successfully!")
```

```python
# Updated final report

print("\n")
print("=" * 50)
print("           PERSONAL EXPENSE REPORT")
print("=" * 50)

print(f"Total Expenses       : PKR {total_expenses:,.2f}")
print(f"Number of Expenses   : {number_of_expenses}")
print(f"Average Expense      : PKR {average_expense:,.2f}")

print("-" * 50)

print(f"Highest Expense      : PKR {highest_expense:,.2f}")
print(f"Highest Expense Type : {highest_expense_category}")

print(f"Lowest Expense       : PKR {lowest_expense:,.2f}")
print(f"Lowest Expense Type  : {lowest_expense_category}")

print("-" * 50)

print(f"Highest Category     : {highest_category}")
print(f"Category Total       : PKR {highest_category_amount:,.2f}")

print(f"Lowest Category      : {lowest_category}")
print(f"Category Total       : PKR {lowest_category_amount:,.2f}")

print("-" * 50)

print("CATEGORY BREAKDOWN")
print("-" * 50)

for category, total in category_totals.items():
    print(f"{category:<20} PKR {total:,.2f}")

print("=" * 50)
```
